# Student Placement Prediction - Dataset Exploratory Data Analysis (EDA)

This notebook performs comprehensive dataset analysis on the student placement prediction dataset.
- **Dataset**: `ml/data/student_placement_prediction_dataset_2026-selected-columns.csv`
- **Target Column**: `placement_status`

```python
import pandas as pd
from pathlib import Path

# Load dataset
data_path = Path('../data/student_placement_prediction_dataset_2026-selected-columns.csv')
df = pd.read_csv(data_path)
print(f"Dataset successfully loaded. Shape: {df.shape}")
```

## 1 & 2. Dataset Shape and Column Names

```python
print(f"Number of Rows: {df.shape[0]:,}")
print(f"Number of Columns: {df.shape[1]}")
print("\nColumn Names:")
for idx, col in enumerate(df.columns, 1):
    print(f"{idx:2d}. {col}")
```

## 3, 9 & 10. Data Types and Column Classification

```python
print("--- Data Types ---")
print(df.dtypes)

numerical_cols = df.select_dtypes(include=['number']).columns.tolist()
categorical_cols = df.select_dtypes(include=['object', 'category', 'string']).columns.tolist()

print(f"\nNumerical Columns ({len(numerical_cols)}): {numerical_cols}")
print(f"Categorical Columns ({len(categorical_cols)}): {categorical_cols}")
```

## 4. Missing / Null Values

```python
missing_df = pd.DataFrame({
    'Missing Count': df.isnull().sum(),
    'Missing Percentage (%)': (df.isnull().sum() / len(df)) * 100
})
missing_df
```

## 5. Duplicate Rows

```python
num_duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {num_duplicates}")
```

## 6. Unique Values of Categorical Columns

```python
for col in categorical_cols:
    print(f"Column '{col}': {df[col].nunique()} unique values -> {list(df[col].unique())}")
```

## 7. Target Column Distribution (`placement_status`)

```python
target_counts = df['placement_status'].value_counts()
target_percents = df['placement_status'].value_counts(normalize=True) * 100
pd.DataFrame({'Count': target_counts, 'Percentage (%)': target_percents.round(2)})
```
