# Student Placement Prediction - Step 5: Model Training and Evaluation

This notebook evaluates 4 classification models using the preprocessed and split data from Steps 3 & 4:
1. **Logistic Regression**
2. **Decision Tree Classifier**
3. **Random Forest Classifier**
4. **Gradient Boosting Classifier**

**Strict Rules Enforced**:
- Models are trained ONLY on `X_train_processed` (80,000 samples).
- Models are evaluated ONLY on `X_test_processed` (20,000 samples).
- `random_state=42` is set across all supported estimators.
- No final model is selected or saved.

```python
import pandas as pd
from pathlib import Path
from model_training import train_and_evaluate_models

data_path = Path('../data/student_placement_prediction_dataset_2026-selected-columns.csv')
eval_results = train_and_evaluate_models(data_path)
comparison_df = eval_results['comparison_df']
cms = eval_results['confusion_matrices']
```

## A & B. Model Comparison Table (Sorted by F1-Score Descending)

```python
comparison_df
```

## C. Confusion Matrices (Test Set: 20,000 samples)

```python
for model_name, cm in cms.items():
    tn, fp, fn, tp = cm.ravel()
    print(f"=== {model_name} ===")
    print(f"  True Negatives  (TN): {tn:,}")
    print(f"  False Positives (FP): {fp:,}")
    print(f"  False Negatives (FN): {fn:,}")
    print(f"  True Positives  (TP): {tp:,}\n")
```

## D. Highest F1-Score Model

```python
best_row = comparison_df.iloc[0]
print(f"Highest F1-Score Model: {best_row['Model']} (F1 = {best_row['F1-score']:.4f})")
```
