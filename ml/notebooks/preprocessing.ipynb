# Student Placement Prediction - Data Preprocessing Demonstration

This notebook demonstrates the preprocessing pipeline built in `ml/notebooks/preprocessing.py`.
- **Target Encoding**: `placement_status` mapped to binary (`Placed` -> 1, `Not Placed` -> 0)
- **Categorical Feature Encoding**: `OneHotEncoder` on `gender`, `branch`, `college_tier`
- **Numerical Features**: Retained as numerical values
- **ColumnTransformer**: Combines numerical and categorical transformations
- **Data Leakage Prevention**: Train/test split performed prior to fitting `ColumnTransformer`
