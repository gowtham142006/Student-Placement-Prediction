# Student Placement Prediction - Step 4: Train/Test Split

This notebook demonstrates Step 4: Stratified 80/20 Train/Test Split & Preprocessing Leakage Prevention.
- **Raw Feature Separation**: Separates raw features `X` and target `y` (`placement_status`).
- **80/20 Stratified Split**: `test_size=0.20`, `random_state=42`, `stratify=y`.
- **Leakage Prevention**: Preprocessor is fitted strictly on `X_train_raw` and only `.transform()` is called on `X_test_raw`.

```python
import pandas as pd
from pathlib import Path
from train_test_split import execute_step_4

data_path = Path('../data/student_placement_prediction_dataset_2026-selected-columns.csv')
data = execute_step_4(data_path)
```

## A. Raw Split Shapes

```python
print(f"X_train shape (raw): {data['X_train_raw'].shape}")
print(f"X_test shape  (raw): {data['X_test_raw'].shape}")
print(f"y_train shape:       {data['y_train'].shape}")
print(f"y_test shape:        {data['y_test'].shape}")
```

## B. Target Distribution (Stratification)

```python
y_train = data['y_train']
y_test = data['y_test']

dist_df = pd.DataFrame({
    'y_train Count': y_train.value_counts(),
    'y_train %': (y_train.value_counts(normalize=True) * 100).round(4),
    'y_test Count': y_test.value_counts(),
    'y_test %': (y_test.value_counts(normalize=True) * 100).round(4)
})
dist_df.index = dist_df.index.map({1: 'Placed (1)', 0: 'Not Placed (0)'})
dist_df
```
